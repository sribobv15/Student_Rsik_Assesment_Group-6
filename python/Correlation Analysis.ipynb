```python

import pandas as pd

# Load the dataset
df = pd.read_csv("Data.csv")

# Descriptive statistics
print(df[["studytime", "failures", "absences", "G3"]].describe())

# Correlation with final grade
correlations = df[["studytime", "failures", "absences", "G3"]].corr()

print(correlations["G3"])

# Compare average final grades
print(df.groupby("failures")["G3"].mean())
print(df.groupby("studytime")["G3"].mean())

# Check missing values
print(df.isnull().sum())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set chart style
sns.set_style("whitegrid")

# Load dataset
df = pd.read_csv("Data.csv")

# Select features
features = ["studytime", "failures", "absences", "G3"]

# Calculate correlations
correlations = df[features].corr()

# Create heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlations,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between Student Performance Factors")
plt.tight_layout()
plt.show()
```
